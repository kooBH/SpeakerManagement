```python
import scipy.io as sio
import librosa as rs
import matplotlib.pyplot as plt
import soundfile as sf
import numpy as np
```

```python


path_doa = "/home/data/IITP/dev/try_5_icecream_0_doa.mat"
path_vad = "/home/data/IITP/dev/try_5_icecream_0_vad.mat"
path_aud = "/home/data/IITP/dev/try_5_icecream_0.wav"

path_doa2vad = "/home/data/IITP/dev/try_5_icecream_0_doa2wav.wav"

doa1 = sio.loadmat(path_doa)
print(doa1.keys())

print(doa1['doa'].shape)

aud1 = rs.load(path_aud,sr=16000)[0]
print(aud1.shape)

plt.figure()
plt.plot(doa1['doa'][0])

vad1 = sio.loadmat(path_vad)
print(vad1['vad'].shape)
#print(vad1)

val = doa1['doa'][0]*vad1['vad'][0]
print(val)

len_o = len(val)
len_n = len(val)*128
arr_o = np.arange(len_o)
arr_n = np.linspace(0,len_o-1, num=len_n)

val = val/180.0
val_n = np.interp(arr_n,arr_o,val)

print(val_n)
len(val_n)

sf.write(path_doa2vad,val_n,16000)
```
